```python
#  order details (customer name, product, price, category) 
```

```python
cust_order_details = [('Mary', 'Shirt', 10.00, 'Clothing'),
                      ('Mary', 'Kindle', 30.00, 'Electronics'),
                      ('Mary', 'Hat', 5.00, 'Clothing'),
                      ('Alessio', 'Table', 50.00, 'Furniture'),
                      ('Alessio', 'Bed', 80.00, 'Furniture'),
                      ('Alessio', 'Chair', 30.00, 'Furniture'),
                      ('Alessio', 'Dresser', 50.00, 'Storage'),
                      ('Alessio', 'Jacket', 40.00, 'Clothing'),
                      ('Alessio', 'Laptop', 80.00, 'Electronics'),
                      ('William', 'Shirt', 10.00, 'Clothing'),
                      ('William', 'Pant', 20.00, 'Clothing'),
                      ('William', 'Water Bottle', 20.00, 'Storage'),
                      ('William', 'Papers', 10.00, 'Stationary'),
                      ('William', 'Marker', 7.00, 'Stationary'),
                      ('Norsen', 'Phone', 90.00, 'Electronics'),
                      ('Norsen', 'Stapler', 20.00, 'Stationary'),
                      ('Norsen', 'Mirror', 20.00, 'Furniture'),
                      ('Pete INC.', 'Air Pods', 67.00, 'Electronics'),
                      ('Pete INC.', 'Lunch Box', 10.00, 'Storage')]
cust_order_details                   
```

```python
#  Dictionary Cust Name : List of Ordered Products
```

```python
from collections import defaultdict

# Create a default dictionary where each value is a list
customer_order_dict = defaultdict(list)

# Add items
for name,product,price,category in cust_order_details:
# my_dict["key"].append("value")
 customer_order_dict[name].append(product)

type(customer_order_dict)
print(customer_order_dict)

# convert to regular dictionary
customer_order_dict = dict(customer_order_dict)
type(customer_order_dict)
print(customer_order_dict)
```

# Task 2: Classify products by category

```python
# Dictionary Product : Category 
# Use a dictionary to map each product to its respective category
```

```python
product_category_dict = defaultdict(set)
for name,product,price,category in cust_order_details:
    product_category_dict[product].add(category)

product_category_dict = dict(product_category_dict)
print(product_category_dict)
```

```python
# set of unique product categories
```

```python
all_categories = product_category_dict.values()
all_categories
```

```python
unique_categories = set()

for categories in product_category_dict.values():
    for category in categories:
        unique_categories.add(category)

print ('Display all available product categories' , unique_categories)
```

# Task 3: Analyze customer orders

```python
# Use a loop to calculate the total amount each customer spends
```

```python
customer_totals = defaultdict(float)

# Loop through each order and accumulate the price
for name, product, price, category in cust_order_details:
    customer_totals[name] += float(price)

# Convert to regular dict if needed
customer_totals = dict(customer_totals)

# Print the result
print(customer_totals)




```

```python
# elif to classify Customer 
```

```python
customer_classification = {}
for name,total in customer_totals.items():
 if int(total)>=100:
    customer_classification[name] = 'High Value Buyer'
 elif int(total)>= 50 and  int(total) <= 100:
    customer_classification[name] = 'Moderate Value Buyer'
 elif int(total) <50:
    customer_classification[name] = 'Low Value Buyer'
 else:
    customer_classification[name] = 'Error on Buyer'

print(customer_classification)
```

# Task 4: Generate business insights

```python
# Dictionary {'Category' : 'Revenue'}
# Calculate the total revenue per product category and store it in a dictionary
```

```python
total_rev_per_product_dict = defaultdict(float)
for name, product, price, category in cust_order_details:
    total_rev_per_product_dict[category] +=float(price)

total_rev_per_product_dict=dict(total_rev_per_product_dict)
print(total_rev_per_product_dict)
    
```

```python
#  Extract unique products from all orders using a set
```

```python
unique_products=set()
for name,product,price,category in cust_order_details:
    if product not in unique_products:
        unique_products.add(product)
print('Unique Products' , unique_products)
```

```python
# Use a list comprehension to find all customers who purchased electronics
```

```python
cust_electronics = set()
for name,product,price,category in cust_order_details:
    if category == 'Electronics':
     cust_electronics.add(name)
print('All customers who purchased electronics' , cust_electronics)
```

```python
# Identify the top three highest-spending customers using sorting
```

```python
# Step 1: Sort by total spent (values), descending
sorted_customers = sorted(customer_totals.items(), key=lambda x: x[1], reverse=True)

# Step 2: Get top 3
top_three = sorted_customers[:3]

# Step 3: Convert to dictionary (optional)
top_three_dict = dict(top_three)

print('top three highest-spending customers', top_three_dict)
```

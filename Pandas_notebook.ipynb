# Series

A Series is very similar to a NumPy array (in fact it is built on top of the NumPy array object). What differentiates the NumPy array from a Series, is that a Series can have axis labels, meaning it can be indexed by a label, instead of just a number location. It also doesn't need to hold numeric data, it can hold any arbitrary Python Object.

## Applying Function on Pandas Column

```python
import pandas as pd
df=pd.read_csv("tips_v1.csv")
df
```

```python
def ch(x):
    x = x.replace('Sun','Sunday')
    return x
```

```python
ch('Sun')
```

```python
def change_name(x):
    
    if x == 'Sun':
        return 'Sunday'
    
    if x == 'Mon':
        return 'Monday'
    
    if x == 'Tue':
        return 'Tuesday'
    
    if x == 'Fri':
        return 'Friday' 
    
    if x == 'Sat':
        return 'Saturday' 
    
    else:
        return x
```

```python
df.head(6)
```

```python
df['day'].apply( change_name )
```

```python
df.head(2)
```

```python
df['day_name'] = df['day'].apply(change_name)
```

```python
df.head(6)
```

```python
df['comb'] = df['gender'] + '_' + df['smoker']
```

```python
df.head(2)
```

```python
'Female_Yes'.split('_')[1]
```

```python
def discount_function(x):
    if x == 'Sunday':
        return 'Discount Allowed'
    else:
        return 'No Discount'

df['Discount'] = df['day_name'].apply(discount_function)
```

```python
df.sample(7)
```

```python
def bucket(x):
    if x >= 0 and x < 10:
        return '10'
    elif x >= 10 and x < 20:
        return '20'
    elif x >= 20 and x < 30:
        return '30'
    elif x >= 30:
        return '30+'

df['total_bill_bucket'] = df['total_bill'].apply(bucket)

df.head()
```

```python
may = pd.DataFrame({'day': [21, 22, 23, 24], 'sales': [100, 200, 110, 160]} , index = [0,1,2,3])

june = pd.DataFrame({'day': [1, 2, 3, 4], 'sales': [210, 180, 190, 96]} , index = [4,5,6,7])

july = pd.DataFrame({'day': [16, 17, 18, 19], 'sales': [160, 114, 110, 150]} , index = [7,8,9,10])
```

```python
pd.concat( [ may , june , july ] , axis = 1 ).fillna('')
```

```python
pd.concat( [ may , june , july ] , axis = 1 , ignore_index = True ).fillna('')
```

```python
yearly_df=pd.DataFrame()

for month_df in [ may , june , july ]:
    
    yearly_df=pd.concat([yearly_df,month_df.reset_index(drop=True)], axis = 1)
    
yearly_df
```
